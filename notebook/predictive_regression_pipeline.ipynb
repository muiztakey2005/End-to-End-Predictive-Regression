```python
# Define features and Target
y=df_clean['SalePrice']
x=df_clean.drop(columns=['SalePrice','Id', 'PID'])
print('Feature Shape:',x.shape)
print('Target Shape:',y.shape)
print('Number of numerical features:',x.select_dtypes(include=['int64','float64']).shape[1])
print('Number of Categorical features:',x.select_dtypes(include=['object']).shape[1])
```

```python
# train-test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print('Training Set:')
print('x_train shape:',x_train.shape)
print('y_train.shape:',y_train.shape)

print('Testing Set:')
print('x_test shape:',x_test.shape)
print('y_test shape:',y_test.shape)
```

```python
#preprocessing pipeline
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer

numeric_features=x_train.select_dtypes(include=['int64','float64']).columns
categorical_features=x_train.select_dtypes(include=['object']).columns

# Numerical preprocessing
numeric_transformer=Pipeline(
    steps=[
        ('imputer',SimpleImputer(strategy='median')),
        ('scaler',StandardScaler())
    ]
)

# Categorical preprocessing
categorical_transformer=Pipeline(
    steps=[
     ('imputer',SimpleImputer(strategy='most_frequent')),
     ('onehot',OneHotEncoder(handle_unknown='ignore'))
    ]
)

# Combine both preprocessing pipeline
preprocessor=ColumnTransformer(
    transformers=[
        ('num',numeric_transformer,numeric_features),
        ('cat',categorical_transformer,categorical_features)
    ]
)
print('Numerical Features:',len(numeric_features))
print('Categorical Features:',len(categorical_features))
```

```python
# Baseline Linear Regression
from sklearn.linear_model import LinearRegression
linear_model=Pipeline(
    steps=[
        ('preprocessor',preprocessor),
        ('regressor',LinearRegression())
    ]
)
linear_model.fit(x_train,y_train)
```

```python
# Model predictions and metrics
from sklearn.metrics import (mean_squared_error,mean_absolute_error,r2_score)
import numpy as np
y_pred=linear_model.predict(x_test)
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test,y_pred)
print('Linear Regression Performance: ')
print(f'MAE: {mae:,.2f}')
print(f'MSE: {mse:,.2f}')
print(f'RMSE: {rmse:,.2f}')
print(f'r2_score: {r2:,.2f}')
```

```python
from sklearn.linear_model import Ridge
ridge_model=Pipeline(
    steps=[
        ('preprocessor',preprocessor),
        ('regressor',Ridge(alpha=10.0))
    ]
)
ridge_model.fit(x_train,y_train)
ridge_pred=ridge_model.predict(x_test)
ridge_mae=mean_absolute_error(y_test,ridge_pred)
ridge_mse=mean_squared_error(y_test,ridge_pred)
ridge_rmse=np.sqrt(ridge_mse)
ridge_r2=r2_score(y_test,ridge_pred)
print('Ridge Regression Performance: ')
print(f'MAE: {ridge_mae:,.2f}')
print(f'MSE: {ridge_mse:,.2f}')
print(f'RMSE: {ridge_rmse:,.2f}')
print(f'r2_score: {ridge_r2:,.2f}')
```

```python
# Random Forest Regression
from sklearn.ensemble import RandomForestRegressor
random_forest_model=Pipeline(
    steps=[
        ('preprocessor',preprocessor),
        ('regressor',(
            RandomForestRegressor(
                n_estimators=300,
                random_state=42,
                n_jobs=-1
            )
        ))
    ]
)
random_forest_model.fit(x_train,y_train)
rf_pred=random_forest_model.predict(x_test)
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_mse = mean_squared_error(y_test, rf_pred)
rf_rmse = np.sqrt(rf_mse)
rf_r2 = r2_score(y_test, rf_pred)

print(f"MAE  : {rf_mae:,.2f}")
print(f"MSE  : {rf_mse:,.2f}")
print(f"RMSE : {rf_rmse:,.2f}")
print(f"r2_score   : {rf_r2:.4f}")
```

```python
# Gradient Boosting Regression
from sklearn.ensemble import GradientBoostingRegressor
gradient_model=Pipeline(
    steps=[
        ('processor',preprocessor),
        ('regressor',GradientBoostingRegressor(
            n_estimators=300,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ))
    ]
)
gradient_model.fit(x_train,y_train)
gb_pred=gradient_model.predict(x_test)
gb_mae = mean_absolute_error(y_test, gb_pred)
gb_mse = mean_squared_error(y_test, gb_pred)
gb_rmse = np.sqrt(gb_mse)
gb_r2 = r2_score(y_test, gb_pred)
print(f"MAE  : {gb_mae:,.2f}")
print(f"MSE  : {gb_mse:,.2f}")
print(f"RMSE : {gb_rmse:,.2f}")
print(f"R²   : {gb_r2:.4f}")
```

```python
# Model Comparison
model_results = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Ridge Regression',
        'Random Forest',
        'Gradient Boosting'
    ],
    'MAE': [
        mae,
        ridge_mae,
        rf_mae,
        gb_mae
    ],
    'MSE': [
        mse,
        ridge_mse,
        rf_mse,
        gb_mse
    ],
    'RMSE': [
        rmse,
        ridge_rmse,
        rf_rmse,
        gb_rmse
    ],
    'R2': [
        r2,
        ridge_r2,
        rf_r2,
        gb_r2
    ]
})
model_results=model_results.sort_values('RMSE',ascending=True).reset_index(drop=True)
print('Model Performance Comparison: ')
display(
    model_results.style.format({
        'MAE': '{:,.2f}',
        'MSE': '{:,.2f}',
        'RMSE': '{:,.2f}',
        'R2': '{:.4f}'
    })
)
```

```python
# Actual vs Predicted
plt.figure(figsize=(8,5))
sns.scatterplot(
    x=y_test,
    y=gb_pred
)
min_value=min(y_test.min(),gb_pred.min())
max_value=max(y_test.max(),gb_pred.max())

plt.plot(
    [min_value,max_value],
    [min_value,max_value],
    linestyle='--',
    color='red'
)
plt.title('Actual vs Predicted SalePrice')
plt.xlabel('Actual SalePrice')
plt.ylabel('Predicted SalePrice')

plt.tight_layout()
plt.show()
```

```python
# Residual Analysis
residuals=y_test-gb_pred
plt.figure(figsize=(8,5))
sns.scatterplot(
    x=gb_pred,
    y=residuals
)

# zero-error refrence line
plt.axhline(
    y=0,
    linestyle='--',
    color='red'
)
plt.title('Residuals vs Predicted SalePrice')
plt.xlabel('Predicted SalePrice')
plt.ylabel('Residuals')

plt.tight_layout()
plt.show()
```

```python
# 5-fold cross validation
from sklearn.model_selection import cross_validate
cv_results=cross_validate(
    gradient_model,
    x_train,
    y_train,
    cv=5,
    scoring={
        'MAE':'neg_mean_absolute_error',
        'RMSE':'neg_root_mean_squared_error',
        'R2':'r2'
    },
    n_jobs=-1
)
# Convert negative error scores to positive values
cv_mae=-cv_results['test_MAE']
cv_rmse=-cv_results['test_RMSE']
cv_r2=cv_results['test_R2']
print('5-Fold Cross Validation Result: ')

print(f'Mean MAE:{cv_mae.mean():,.2f}')
print(f'Std MAE:{cv_mae.std():,.2f}')

print(f'Mean RMSE:{cv_rmse.mean():,.2f}')
print(f'Std MAE:{cv_rmse.std():,.2f}')

print(f'Mean R2:{cv_r2.mean():,.2f}')
print(f'Std R2:{cv_r2.std():,.2f}')
```

```python
# Hyperparameter Tuning
from sklearn.model_selection import GridSearchCV
param_grid={
    'regressor__n_estimators':[200,300,400],
    'regressor__learning_rate':[0.03,0.05,0.08],
    'regressor__max_depth':[2,3,4]
}

grid_search=GridSearchCV(
    estimator=gradient_model,
    param_grid=param_grid,
    cv=5,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1,
    verbose=1
)
grid_search.fit(x_train,y_train)
print('Best Parameters:')
print(grid_search.best_params_)

print("Best CV RMSE:")
print(f"{-grid_search.best_score_:,.2f}")
```

```python
# Final Tuned Model Evaluation
# best model found through Gridsearchcv
best_model=grid_search.best_estimator_

best_pred=best_model.predict(x_test)

final_mae = mean_absolute_error(y_test, best_pred)
final_mse = mean_squared_error(y_test, best_pred)
final_rmse = np.sqrt(final_mse)
final_r2 = r2_score(y_test, best_pred)

print(f"MAE  : {final_mae:,.2f}")
print(f"MSE  : {final_mse:,.2f}")
print(f"RMSE : {final_rmse:,.2f}")
print(f"R2   : {final_r2:.4f}")

```

```python
final_results = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Ridge Regression',
        'Random Forest',
        'Gradient Boosting',
        'Tuned Gradient Boosting'
    ],
    'MAE': [
        mae,
        ridge_mae,
        rf_mae,
        gb_mae,
        final_mae
    ],
    'MSE': [
        mse,
        ridge_mse,
        rf_mse,
        gb_mse,
        final_mse
    ],
    'RMSE': [
        rmse,
        ridge_rmse,
        rf_rmse,
        gb_rmse,
        final_rmse
    ],
    'R2': [
        r2,
        ridge_r2,
        rf_r2,
        gb_r2,
        final_r2
    ]
})

final_results = final_results.sort_values(
    by='RMSE',
    ascending=True
).reset_index(drop=True)

display(
    final_results.style.format({
        'MAE': '{:,.2f}',
        'MSE': '{:,.2f}',
        'RMSE': '{:,.2f}',
        'R2': '{:.4f}'
    })
)
```
